```python
import numpy as np
import matplotlib.pyplot as plt

# Input feature: hours studied
X = np.array([1, 2, 3, 4, 5], dtype=float)

# Actual exam scores
y = np.array([30, 40, 50, 60, 70], dtype=float)

# Display the data
plt.scatter(X, y)
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.title("Hours Studied vs Exam Score")
plt.show()
```

```python
theta_0 = 0.0  # Bias
theta_1 = 0.0  # Weight

learning_rate = 0.01
epochs = 2000
m = len(X)
```

```python

# Store the cost at every epoch so we can plot learning
cost_history = []

for epoch in range(epochs):

    # STEP 1: Make predictions using current parameters
    y_pred = theta_0 + theta_1 * X

    # STEP 2: Calculate the errors
    errors = y_pred - y

    # STEP 3: Calculate the cost (Mean Squared Error)
    mse = np.mean(errors ** 2)
    cost_history.append(mse)

    # STEP 4: Calculate the gradients
    gradient_theta_0 = (2 / m) * np.sum(errors)

    gradient_theta_1 = (2 / m) * np.sum(errors * X)

    # STEP 5: Update the bias and weight
    theta_0 = theta_0 - learning_rate * gradient_theta_0

    theta_1 = theta_1 - learning_rate * gradient_theta_1

# Display the learned parameters
print("Learned bias:", theta_0)
print("Learned weight:", theta_1)
print("Final MSE:", cost_history[-1])
print("Final RMSE:", np.sqrt(cost_history[-1]))
```

```python
plt.plot(cost_history)
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.title("Gradient Descent: Cost During Training")
plt.show()
```

```python
y_pred = theta_0 + theta_1 * X

plt.scatter(X, y, label="Actual data")
plt.plot(X, y_pred, label="Learned regression line")
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.title("Linear Regression Trained with Gradient Descent")
plt.legend()
plt.show()
```
